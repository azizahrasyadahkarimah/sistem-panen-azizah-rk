```python
import os

# 1. PERSIAPAN FOLDER PROYEK
print("=== 1. MEMBUAT FOLDER 'sistem-panen' ===")
os.makedirs("sistem-panen", exist_ok=True)
os.chdir("sistem-panen")


# 2. LANGKAH A: INISIALISASI GIT & COMMIT PERTAMA
print("\n=== LANGKAH A: INISIALISASI & COMMIT PERTAMA ===")
# Inisialisasi konfigurasi Git lokal
!git config --global user.name "Mahasiswa TIN"
!git config --global user.email "mahasiswa@apps.ipb.ac.id"

# Inisialisasi repositori Git
!git init

# Buat berkas panen.py awal
kode_awal = """def hitung_total(hasil_kg, harga_per_kg):
    return hasil_kg * harga_per_kg

print("Total Panen:", hitung_total(100, 5000))
"""
with open("panen.py", "w") as f:
    f.write(kode_awal)

# Tambahkan file ke staging area dan lakukan commit pertama
!git add panen.py
!git commit -m "Initial commit: Menambahkan program penghitung total hasil panen"


# 3. LANGKAH B: BUAT BRANCH 'fitur-diskon' & COMMIT PERUBAHAN
print("\n=== LANGKAH B: BRANCH BARU & COMMIT FITUR DISKON ===")
# Buat dan berpindah ke branch 'fitur-diskon'
!git checkout -b fitur-diskon

# Tambahkan fungsi diskon ke berkas panen.py
kode_dengan_diskon = kode_awal + """
def hitung_diskon(total, persen_diskon):
    return total * (persen_diskon / 100)

print("Diskon (10%):", hitung_diskon(500000, 10))
"""
with open("panen.py", "w") as f:
    f.write(kode_dengan_diskon)

# Commit perubahan di branch 'fitur-diskon'
!git add panen.py
!git commit -m "Feat: Menambahkan fungsi perhitungan diskon panen"


# 4. LANGKAH C: KEMBALI KE MAIN & MERGE
print("\n=== LANGKAH C: KEMBALI KE MAIN & MERGE ===")
# Kembali ke branch utama (main)
!git checkout main

# Lakukan merge dari branch 'fitur-diskon'
!git merge fitur-diskon


# 5. RIWAYAT COMMIT AKHIR (GIT LOG)
print("\n=== RIWAYAT COMMIT AKHIR (GIT LOG) ===")
!git log --oneline
```

```python
import os

# =====================================================================
# 1. KONFIGURASI AUTENTIKASI GITHUB
# =====================================================================
GITHUB_USERNAME = "USERNAME_GITHUB_ANDA"  # Ganti dengan username GitHub Anda
GITHUB_TOKEN = "PAT_TOKEN_GITHUB_ANDA"   # Ganti dengan Personal Access Token GitHub
REPO_NAME = "sistem-panen-timi"          # Nama repositori GitHub Anda

# URL Remote Repository dengan autentikasi Token
REMOTE_URL = f"https://{GITHUB_USERNAME}:{GITHUB_TOKEN}@github.com/{GITHUB_USERNAME}/{REPO_NAME}.git"

# Set identitas Git
!git config --global user.name "Timi"
!git config --global user.email "email_anda@apps.ipb.ac.id"  # Ganti dengan email Anda


# =====================================================================
# 2. INISIALISASI PROYEK LOKAL & GIT
# =====================================================================
print("=== 1. MEMBUAT PROYEK LOKAL & INISIALISASI GIT ===")
os.makedirs(REPO_NAME, exist_ok=True)
os.chdir(REPO_NAME)

!git init
!git branch -M main

# Buat berkas panen.py awal (Poin 2a)
kode_awal = """def hitung_total(hasil_kg, harga_per_kg):
    return hasil_kg * harga_per_kg

print("Total Panen:", hitung_total(100, 5000))
"""
with open("panen.py", "w") as f:
    f.write(kode_awal)

# Commit Pertama
!git add panen.py
!git commit -m "Initial commit: Menambahkan program penghitung total hasil panen"


# =====================================================================
# 3. MEMBUAT BRANCH BARU & MERGE (Poin 2b & 2c)
# =====================================================================
print("\n=== 2. MEMBUAT BRANCH 'fitur-diskon' & MERGE ===")
# Buat dan pindah ke branch fitur-diskon
!git checkout -b fitur-diskon

# Tambahkan fungsi diskon
kode_dengan_diskon = kode_awal + """
def hitung_diskon(total, persen_diskon):
    return total * (persen_diskon / 100)

print("Diskon (10%):", hitung_diskon(500000, 10))
"""
with open("panen.py", "w") as f:
    f.write(kode_dengan_diskon)

# Commit di branch fitur-diskon
!git add panen.py
!git commit -m "Feat: Menambahkan fungsi perhitungan diskon panen"

# Kembali ke branch main dan lakukan merge
!git checkout main
!git merge fitur-diskon


# =====================================================================
# 4. SUNTING README.md & UNGGAH KE GITHUB (Poin 3a & 3b)
# =====================================================================
print("\n=== 3. MEMBUAT README.md & PUSH KE GITHUB ===")

# Menulis teks README.md menggunakan penggabungan string yang rapi
url_clone = "https://github.com/" + GITHUB_USERNAME + "/" + REPO_NAME + ".git"

isi_readme = (
    "# Sistem Pencatatan Hasil Panen Digital\n\n"
    "## Deskripsi Singkat\n"
    "Sistem Pencatatan Hasil Panen Digital adalah program Python sederhana "
    "yang dirancang untuk mengotomatisasi perhitungan total pendapatan panen "
    "serta kalkulasi diskon penjualan produk agroindustri secara cepat dan terstruktur.\n\n"
    "## Cara Menjalankan Program\n"
    "1. Pastikan Python 3.x telah terinstal di perangkat Anda.\n"
    "2. Unduh atau klon repositori ini:\n"
    "   git clone " + url_clone + "\n"
    "3. Masuk ke direktori proyek:\n"
    "   cd " + REPO_NAME + "\n"
    "4. Jalankan berkas utama:\n"
    "   python panen.py\n"
)

with open("README.md", "w") as f:
    f.write(isi_readme)

# Commit README.md
!git add README.md
!git commit -m "Docs: Menyunting README.md dengan deskripsi dan petunjuk penggunaan"

# Hubungkan ke remote repository dan Push ke GitHub
!git remote remove origin 2>/dev/null
!git remote add origin {REMOTE_URL}
!git push -u origin main


# =====================================================================
# 5. MENAMPILKAN RIWAYAT COMMIT
# =====================================================================
print("\n=== 4. RIWAYAT COMMIT AKHIR (GIT LOG) ===")
!git log --oneline
```
